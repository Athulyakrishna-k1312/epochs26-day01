```python
plt.figure(figsize=(8,5))
sns.histplot(df["popularity"], bins=20)
plt.title("Distribution of Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Songs")
plt.show()
```

## Visualization 1

Insight:
Most songs have low to medium popularity, while very popular songs are less common.

```python
plt.figure(figsize=(10,5))
sns.countplot(
    y="track_genre",
    data=df,
    order=df["track_genre"].value_counts().head(10).index
)
plt.title("Top 10 Genres")
plt.show()
```

## Visualization 2

Insight:
Some genres have many more tracks than others in the dataset.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x="danceability",
    y="energy",
    data=df
)
plt.title("Danceability vs Energy")
plt.show()
```

## Visualization 3

Insight:
Songs with higher danceability often have moderate to high energy.

```python
plt.figure(figsize=(6,5))
sns.boxplot(
    x="explicit",
    y="popularity",
    data=df
)
plt.title("Popularity by Explicit Content")
plt.show()
```

## Visualization 4

Insight:
Explicit and non-explicit songs have overlapping popularity distributions.

```python
plt.figure(figsize=(10,8))
sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()
```

## Visualization 5

Insight:
Some numerical features show positive or negative relationships, while many are only weakly correlated.

```python
top_artists = df["artists"].value_counts().head(10)

plt.figure(figsize=(10,5))
top_artists.plot(kind="bar")
plt.title("Top 10 Artists")
plt.xlabel("Artist")
plt.ylabel("Number of Tracks")
plt.show()
```

## Visualization 6

Insight:
A small number of artists contribute a large number of tracks to the dataset.
